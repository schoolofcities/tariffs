# Create Tariff Impacts Excel Files

Creates `tariff-impacts-ada-data.xlsx` and `tariff-impacts-csd-data.xlsx` with:
- **Counts** sheet: Business/Employee counts per tariff category
- **Percents** sheet: Percentage impacts per tariff category

```python
import pandas as pd
import geopandas as gpd
from datetime import datetime
import os

os.environ['OGR_GEOJSON_MAX_OBJ_SIZE'] = '0'
```

```python
# ADA DATA

# Load ADA counts from centroids.geojson (has geometry) 
# and percents from choropleth.geojson (has geometry)
ada_counts = gpd.read_file('centroids.geojson')
ada_percents = gpd.read_file('choropleth.geojson')

# Convert geometry to WKT string for Excel compatibility
ada_counts['geometry'] = ada_counts['geometry'].apply(lambda x: x.wkt if x else None)
#ada_percents['geometry'] = ada_percents['geometry'].apply(lambda x: x.wkt if x else None)

# Create Excel file with both sheets
with pd.ExcelWriter('tariff-impacts-ada-data-1-26.xlsx', engine='openpyxl') as writer:
    ada_counts.to_excel(writer, sheet_name='Counts', index=False)
    ada_percents.to_excel(writer, sheet_name='Percents', index=False)

print(f"Created tariff-impacts-ada-data-1-26.xlsx")
print(f"   Counts: {ada_counts.shape[0]} rows, {ada_counts.shape[1]} columns")
print(f"   Percents: {ada_percents.shape[0]} rows, {ada_percents.shape[1]} columns")
```

```python
# CSD DATA

# Load CSD counts from centroids_csd.geojson and percents from choropleth_csd.geojson
csd_counts = gpd.read_file('centroids_csd.geojson')
csd_percents = gpd.read_file('choropleth_csd.geojson')

# Convert geometry to WKT string for Excel compatibility
csd_counts['geometry'] = csd_counts['geometry'].apply(lambda x: x.wkt if x else None)
#csd_percents['geometry'] = csd_percents['geometry'].apply(lambda x: x.wkt if x else None)

# Create Excel file with both sheets
with pd.ExcelWriter('tariff-impacts-csd-data-1-26.xlsx', engine='openpyxl') as writer:
    csd_counts.to_excel(writer, sheet_name='Counts', index=False)
    csd_percents.to_excel(writer, sheet_name='Percents', index=False)

print(f"Created tariff-impacts-csd-data-1-26.xlsx")
print(f"   Counts: {csd_counts.shape[0]} rows, {csd_counts.shape[1]} columns")
print(f"   Percents: {csd_percents.shape[0]} rows, {csd_percents.shape[1]} columns")
```
